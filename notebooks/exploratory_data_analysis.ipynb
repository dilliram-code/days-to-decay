# DAYS TO DECAY: Exploratory Data Analysis & Target Distribution
## Visual Deterioration & Continuous Shelf-Life Regression

This notebook explores the empirical distribution of perishable food aging metrics, group-based sample distributions, target shelf-life variables (`remaining_days`), and visual condition transitions.

```python
import os
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
metadata_path = Path("../data/metadata/dataset.csv")

if metadata_path.exists() and metadata_path.stat().st_size > 0:
    df = pd.read_csv(metadata_path)
    print(f"Loaded dataset with {len(df)} records.")
    display(df.head())
else:
    print("dataset.csv not populated yet. Follow DATASET_GUIDE.md to collect photographs.")
```

### 1. Perishable Category Distribution & Unique Physical Samples

```python
if 'df' in locals() and not df.empty:
    plt.figure(figsize=(8, 4), dpi=150)
    sns.countplot(data=df, x='food_category', palette='viridis')
    plt.title("Photographs per Food Category")
    plt.xlabel("Category")
    plt.ylabel("Count")
    plt.show()
    
    samples_per_cat = df.groupby('food_category')['sample_id'].nunique()
    print("Unique Physical Samples per Category:")
    print(samples_per_cat)
```

### 2. Regression Target Distribution: `remaining_days`

Checking skewness, spread, and natural limits. Shelf life is strictly non-negative: $\text{remaining\_days} \ge 0.0$.

```python
if 'df' in locals() and not df.empty and 'remaining_days' in df.columns:
    plt.figure(figsize=(9, 4), dpi=150)
    sns.histplot(data=df, x='remaining_days', hue='food_category', kde=True, bins=15)
    plt.title("Distribution of Remaining Shelf Life (Days)")
    plt.xlabel("Remaining Days until Spoilage Endpoint")
    plt.show()
```

### 3. Aging Trajectory: Elapsed Capture Day vs. Remaining Shelf Life

Each physical sample follows an inverse linear decay curve: $\text{remaining\_days} = \text{total\_shelf\_life} - \text{elapsed\_day}$.

```python
if 'df' in locals() and not df.empty and 'capture_day' in df.columns:
    plt.figure(figsize=(9, 5), dpi=150)
    sns.lineplot(data=df, x='capture_day', y='remaining_days', hue='food_category', style='sample_id', markers=True)
    plt.title("Aging Trajectories across Physical Food Samples")
    plt.xlabel("Elapsed Day from Collection (Day 0, 1, 2...)")
    plt.ylabel("Remaining Days")
    plt.show()
```
